# Projeto Final — Aplicação de Sistemas Multirrobôs

## Enunciado

Com base nos conceitos estudados sobre a taxonomia de sistemas multirrobôs, desenvolva um cenário original, real ou hipotético, no qual múltiplos robôs executem uma missão conjunta.

O cenário poderá envolver, por exemplo, aplicações na agricultura, indústria, logística, inspeção, monitoramento ambiental, exploração, segurança, saúde, busca e salvamento ou cidades inteligentes.

O mesmo cenário deverá ser utilizado em todas as questões, garantindo coerência entre as características do sistema e as decisões de projeto.


## Criação de imagens com Inteligência Artificial

Utilize uma ferramenta de Inteligência Artificial generativa para criar uma ou mais imagens representativas da aplicação proposta.

As imagens deverão:
- representar claramente o cenário e a missão;
- mostrar os diferentes robôs que compõem o sistema;
- evidenciar, sempre que possível, a interação entre os robôs;
- estar acompanhadas de uma breve legenda explicativa;
- ser inseridas no arquivo entregue.

Informe também:
- a ferramenta de IA utilizada;
- o comando ou prompt empregado na geração;
- eventuais ajustes realizados posteriormente.

A imagem tem finalidade ilustrativa e não substitui as justificativas técnicas solicitadas.


## Questões
### Questão 1 — Benevolência × Competitividade
Descreva inicialmente o cenário de aplicação e a missão conjunta a ser realizada.

Em seguida, explique se os robôs compartilham um objetivo comum, caracterizando um sistema benevolente, ou se competem por metas, tarefas ou recursos. Justifique tecnicamente sua classificação.

### Questão 2 — Homogêneo × Heterogêneo
Classifique o sistema como homogêneo ou heterogêneo.

Descreva brevemente os robôs que compõem a equipe, considerando sua estrutura física, mobilidade, sensores, atuadores e funções. Explique por que essa composição é adequada à missão proposta.

### Questão 3 — Tamanho do grupo
Informe a quantidade aproximada de robôs empregada na aplicação e classifique o grupo como pequeno — até algumas dezenas de unidades — ou grande — centenas ou mais.

Justifique a quantidade escolhida considerando a extensão do ambiente, a complexidade da missão, o tempo disponível, os custos e a necessidade de cobertura ou redundância.

### Questão 4 — Distribuição do controle
Classifique a arquitetura de controle como centralizada, distribuída ou hierárquica.

Explique como as decisões são tomadas, como as tarefas são atribuídas e como as informações circulam entre os robôs. Justifique por que essa arquitetura é adequada ao cenário, considerando comunicação, escalabilidade, robustez e tolerância a falhas.

### Questão 5 — Tipo de interação
Classifique a interação predominante entre os robôs como coordenação, cooperação ou colaboração.

Apresente exemplos concretos de como essa interação ocorre durante a missão. Explique quais informações, recursos ou tarefas são compartilhados e como a atuação conjunta contribui para alcançar o objetivo do sistema.


## Orientações para entrega
O trabalho deverá ser entregue em um único arquivo PDF, contendo:

1. título e identificação do estudante;
2. descrição geral do cenário;
3. imagem ou imagens produzidas com auxílio de IA;
4. identificação da ferramenta e do prompt utilizado;
5. enunciado completo de cada questão;
6. resposta apresentada imediatamente após a respectiva questão;
7. referências consultadas, quando aplicável.

As **perguntas e as respostas deverão estar claramente destacadas**, utilizando títulos, negrito, cores ou outra formatação que permita diferenciá-las facilmente.

O arquivo deverá apresentar organização visual, linguagem técnica, respostas objetivas e coerência entre todas as dimensões da taxonomia analisadas.


## Critérios de avaliação
- originalidade e clareza do cenário proposto;
- qualidade e pertinência das imagens produzidas;
- coerência entre as respostas e o conteúdo do trabalho;
- aplicação correta dos conceitos de sistemas multirrobôs;
- coerência entre as classificações adotadas;
- qualidade das justificativas técnicas;
- apresentação, organização e legibilidade do arquivo;
- indicação adequada da ferramenta de IA e do prompt utilizado.
