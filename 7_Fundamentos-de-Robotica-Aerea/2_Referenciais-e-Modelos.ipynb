# Referenciais e Modelos

## Sistemas de Referência
- Referencial Global (Inercial ou do Mundo)
- Referencial Espacial (paralelo ao Global)
- Referencial do Corpo (normalmente com origem no centro de gravidade - CoG - do Robô)

## Noções de Modelagem
- Newton-Euler: Referencial no corpo do robô
- Euler-Lagrange: Referencial no mundo

## Estratégias de Controladores de Voo
- Laços internos e externos
- Controle acoplado
- Controle inteligente

## Controladores de Voo
- Controle de Altitude e Guinada
- Controle PVTOL (Planar Vertical Taking off and Landing)
- Controle cinemático e/ou dinâmico
- Controle comutado
