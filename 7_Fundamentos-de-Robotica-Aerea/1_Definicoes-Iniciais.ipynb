## Classificação dos Veículos Aéreos
- Mais leves que o ar: Balões, dirigíveis
- Mais pesados que o ar: asas rotativas (helicópteros, drones), asas fixas (aviões, planadores) e bird-like

## Comparação entre Plataformas Aéreas
Critérios: Consumo de energia, Controlabilidade, Voo estacionário, Voo a baixas velocidades, Voo a altas velocidades, Miniaturização, Decolagem Vertical, Utilização Indoor.

## Desafios da Robótica Aérea
- Navegação: mapeamento e representação do ambiente
- Reconhecimento: percepção do entorno
- Aprendizado: adaptação a condições variáveis
- Cooperação: execução de tarefas em conjunto
- Cognição: uso de conhecimento na tomada de decisão
