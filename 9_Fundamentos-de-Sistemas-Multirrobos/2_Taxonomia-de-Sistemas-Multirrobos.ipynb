# 2. Taxonomia de Sistemas Multirrobôs

## Benevolência × Competitividade
- Os robôs podem atuar de forma colaborativa (benevolente) ou concorrente (competitiva), dependendo dos objetivos individuais e coletivos.

## Homogêneo × Heterogêneo
- Sistemas com robôs idênticos (hardware/software) ou com diferentes capacidades e funções.

## Tamanho
- Desde grupos pequenos com poucos robôs até enxames com centenas ou milhares de unidades.

## Distribuição de Controle
- Pode ser centralizado (um nó de controle principal) ou distribuído (cada robô toma decisões localmente).
