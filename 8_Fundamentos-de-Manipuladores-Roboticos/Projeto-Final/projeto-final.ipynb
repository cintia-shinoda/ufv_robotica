# Atividade Prática – Cinemática Direta e Inversa de Manipuladores no CopeliaSim

## Descrição Geral:
Nesta atividade, o estudante deverá desenvolver um código de controle para um manipulador robótico, com o objetivo de gerar uma trajetória linear no espaço cartesiano, partindo de um ponto A até um ponto B, por meio das cinemáticas direta e inversa para especificar o movimento do robô e um tipo de manipulador definido conforme as regras abaixo. O robô será simulado no ambiente CopeliaSim.

1. Definição do Manipulador:

A seleção será feita conforme a seguinte regra:

- Considere sua matrícula composta por seis dígitos: RSTXYZ.
- Calcule: Resto de divisão inteira da matrícula por 3: Matrícula % 3.
- Resultado:
    - 0 → SCARA
    - 1 → Cilíndrico
    - 2 → Esférico ou Polar

2. Pontos de Partida e Chegada:
- Partida: [-0.5, -0.5, 0.5] m.
- Chegada: [0.5, 0.5, 1.5] m.
- Para cada componente, se o dígito correspondente for par, soma-se 0.25, caso contrário, subtrai-se 0.25.
- NOTA: Caso o ponto de partida, de chegada ou mesmo algum ponto da trajetória de referência apresente alguma restrição quando o movimento do efetuador, por exemplo, colisão com o corpo do robô ou se localizar fora do espaço de trabalho do manipulador, faça a adequação necessária e informe o motivo da alteração. 

Exemplo:

Matrícula: 072943

- Ponto A = [-0.5, -0.5, 0.5] + [0.25, -0.25, 0.25] → [-0.25, -0.75, 0.75]
- Ponto B = [0.5, 0.5, 1.5] + [-0.25, 0.25, -0.25] → [0.25, 0.75, 1.25]


3. Tarefas do Estudante:
- Calcular o alcance máximo e mínimo do manipulador, representando graficamente.
- Implementar:
    - Cinemática direta (parâmetros DH).
    - Cinemática inversa (formalismo fornecido).
- Gerar referências de movimento.
- Executar o robô no CopeliaSim (ou em outro simulador que achar interessante implementar - AuRoRA, Gazebo, Webots, etc.).
- Gerar gráficos:
    - Posição do efetuador ao longo do tempo.
    - Velocidade do efetuador ao longo do tempo.
    - Posição angular ou deslocamento linear das juntas.
- Produzir vídeo explicativo (até 3 minutos).


4. Recursos Fornecidos:

- Modelos de robôs prontos.
- Formalismo da cinemática inversa.

7. Informações Úteis:
- Cinemática Inversa dos manipuladores 
    - Esférico
        - $\theta_1 = arctan(\frac{y}{x})$
        - $\theta_2 = arcos(\frac{z}{r})
        - $d_3 = r = L_1$

 onde "r" é o raio da esfera
 
    - Cilíndrico
        - $\theta_1 = arctan(\frac{y}{x})$
        - $d_2 = z - L_1$
        - $d_3 = r - L_2$

  onde  "r" é o raio do círculo
 
    - Scara
        - $d_3 = -L_0 + z$  
        - $\theta_2 = arccos (\frac{x^2 + y^2- L^2_1 - L^2_2}{2L_1L_2})$
        - $\theta_1 = arctan(\frac{y}{x}) - \arctan(\frac{L_2\sin(\theta_2)}{L_1 + L_2\cos(\theta_2)})$

    - Dimensões dos elos dos robôs:
      - Esférico: L1 = 70cm
      - Cilíndrico: L1 = 70cm, L2 = 70cm
      - Scara: L0 = 120cm, L1 = 60cm, L2 = 60cm
